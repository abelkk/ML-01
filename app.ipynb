```python

import pandas as pd 
import numpy as np
import seaborn as sns # visualizations
from sklearn.datasets import load_wine
from sklearn.ensemble import RandomForestClassifier # cassification problem
from sklearn.model_selection import train_test_split # we split the data to use it well 
from sklearn.metrics import r2_score, accuracy_score, classification_report,confusion_matrix # evaluation metric  - R^2 
import matplotlib.pyplot as plt # aids in visualization
from sklearn.tree import DecisionTreeClassifier
```

#### **Load data**

```python
data =load_wine()
```

```python
data
```

```python
X = pd.DataFrame(data.data, columns=data.feature_names)
y = data.target
```

```python
X.head(2)
```

```python
print(np.unique(y, return_counts=True))
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.4, random_state=42, stratify=y)
print(X_train.shape)
```

## **Modelling**

Expression of variance in dt

```python
dt_model = DecisionTreeClassifier(random_state=42)
dt_model.fit(X_train, y_train)
y_dt_predict = dt_model.predict(X_test)
print(f"accuracy score: {accuracy_score(y_test, y_dt_predict)}")
```

```python
rf_model =RandomForestClassifier(
   n_estimators=100,#number of trees in the forest
   max_features='sqrt', # number of features to consider at every split
   max_depth=6, # maximum number of levels in tree)
   bootstrap=True,#settles on bagging approach hence samplei picked with replacement 
   oob_score=False,
   random_state=42,
   n_jobs=None,
)
```

```python
rf_model.fit(X_train, y_train)
rf_y_pred =rf_model.predict(X_test)
```

```python
accuracy_score(y_test, rf_y_pred)
print(f"accuracy score: {accuracy_score(y_test, rf_y_pred)}")
```
